## Modellen

For en hopper med masse $m$ virker tyngdekraften $mg$ nedover og luftmotstanden $L = kv^2$ oppover. Med positiv retning nedover gir Newtons 2. lov

$$a = \frac{mg - kv^2}{m} = g - \frac{k}{m}v^2$$

Konstanten $k$ avhenger av lufttettheten, formen og arealet. For en hopper i fritt fall er $k \approx 0{,}25$ kg/m.

**Terminalfarten** $v_T$ er farten der luftmotstanden er like stor som tyngdekraften, slik at $a = 0$.
